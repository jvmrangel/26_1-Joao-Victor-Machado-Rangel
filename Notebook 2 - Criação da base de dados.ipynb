```python
# prompt: Quero que leia o arquivo dados.csv

import pandas as pd

# Certifique-se de que o arquivo 'dados.csv' esteja no mesmo diretório do notebook ou forneça o caminho completo.

df = pd.read_csv('respostasForms.csv')
print(df.head()) # Exibe as primeiras linhas do DataFrame para verificar se a leitura foi bem-sucedida
```

```python
df_de_para = pd.read_excel('de_para.xlsx')
print(df_de_para.head()) # Exibe as primeiras linhas do DataFrame para verificar se a leitura foi bem-sucedida
```

```python
# prompt: quero analise a df anterior, terá uma coluna nesse dataset chamada "Selecione uma lista para preencher, de preferencia o primeiro item da lista para garantir a aleatoriedade na metodologia.", é a oitava coluna. Com essa coluna quero que faça um filtro removendo toda ocorrência de resposta igual após a 5 aparição, levando sempre em consideração a ordem de registro da linha (informação encontrada na primeira coluna "Carimbo de data/hora"). Esclarecendo, por exemplo, se na linha 25 temos a aparição do valor "Lista 5" na coluna analisada pela quinta vez, todas as próximas aparições com esse mesmo valor será removido do dataset, por ordem de preenchimento.

# Mapear os nomes das colunas para facilitar o acesso
coluna_selecao = "Selecione uma lista para preencher, de preferencia o primeiro item da lista para garantir a aleatoriedade na metodologia."
coluna_timestamp = "Carimbo de data/hora"

# Converter a coluna de timestamp para datetime para garantir a ordem correta
df[coluna_timestamp] = pd.to_datetime(df[coluna_timestamp])

# Ordenar o DataFrame pelo timestamp
df = df.sort_values(by=coluna_timestamp)

# Dicionário para contar a ocorrência de cada valor na coluna de seleção
contagem_valores = {}

# Lista para armazenar os índices das linhas a serem removidas
indices_para_remover = []

# Iterar sobre as linhas do DataFrame
for index, row in df.iterrows():
    valor_selecao = row[coluna_selecao]

    # Incrementar a contagem do valor
    contagem_valores[valor_selecao] = contagem_valores.get(valor_selecao, 0) + 1

    # Se a contagem for maior que 5, adicionar o índice para remoção
    if contagem_valores[valor_selecao] > 5:
        indices_para_remover.append(index)

# Remover as linhas com base nos índices coletados
df_filtrado = df.drop(indices_para_remover)

# Exibir as primeiras linhas do DataFrame filtrado para verificar o resultado
print("\nDataFrame após o filtro:")
print(df_filtrado.head())

# Opcional: Você pode querer resetar o índice do novo DataFrame
df_filtrado = df_filtrado.reset_index(drop=True)
print("\nDataFrame após resetar o índice:")
print(df_filtrado.head())
```

```python
# prompt: quero que faça uma contagem de cada valor distintos na coluna "Selecione uma lista para preencher, de preferencia o primeiro item da lista para garantir a aleatoriedade na metodologia." dentro do dataset "df_filtrado"

# Realizar a contagem dos valores distintos na coluna especificada do DataFrame filtrado
contagem_valores_distintos = df_filtrado[coluna_selecao].value_counts()

# Exibir a contagem dos valores distintos
print("\nContagem de cada valor distinto na coluna 'Selecione uma lista para preencher...':")
contagem_valores_distintos
```

```python
import pandas as pd

# Transforme a segunda base para que cada linha represente um "Tweet" da coluna 9 em diante
df_melted = pd.melt(df_filtrado, id_vars=df_filtrado.columns[:8], var_name='Tweet', value_name='resposta')


# Junte as duas bases de dados com base na coluna "Tweet"
df_final = pd.merge(df_de_para, df_melted, on='Tweet', how='left')


# Reorganize o dataframe final para ter as colunas desejadas: Palavra, Tweet, resposta
#df_final = df_melted[['Palavra', 'Tweet', 'resposta']]

# Salve o dataframe final em um novo arquivo CSV
#df_final.to_csv('terceira_base.csv', index=False)

df_final = df_melted[['Tweet', 'resposta']]
df_final.to_csv('terceira_base.csv', index=False)

print(df_final.head())
```

```python
# prompt: print o volume de valores distintos dentro da coluna "Tweet" no dataset "df_final"

print(f"O volume de valores distintos na coluna 'Tweet' é: {df_final['Tweet'].nunique()}")
```

```python

df_final_sem_nulos = df_final.dropna(subset=['resposta'])

print(df_final_sem_nulos.head())

print(f"O volume de valores distintos na coluna 'Tweet' é: {df_final_sem_nulos['Tweet'].nunique()}")

# Salve o DataFrame sem valores nulos em um novo arquivo CSV (opcional)
#df_final_sem_nulos.to_csv('terceira_base_sem_nulos.csv', index=False)
```

```python

#tweet_resposta_counts = df_final_sem_nulos.groupby(['Tweet', 'resposta']).size().reset_index(name='counts')
tweet_resposta_counts = df_final.groupby(['Tweet', 'resposta']).size().reset_index(name='counts')

# Exiba a estatística
print(tweet_resposta_counts)

# Salve o DataFrame com as contagens em um novo arquivo CSV (opcional)
tweet_resposta_counts.to_csv('tweet_resposta_counts.csv', index=False)
```

```python
# prompt: Dado o dataset "tweet_resposta_counts" com 3 colunas, sendo elas "Tweet", "resposta", "counts", quero que gere uma tabela final com apenas 2 colunas, uma sendo "Tweet" e a outra sendo a coluna resposta. Porém para montar essa regra, vamos classificar cada Tweet baseado nas outras duas colunas. Quero que faça uma classificação onde na situação numero um resposta = "Considero misógina" e counts >= 3 Então retorne  "É Misógina", qual quer resultado diferente disto, retorne "Não misógina". Desta forma esperase uma tabela final com um resultado para cada tweet, o nome das colunas podem ser "Tweet" e "classificação"

def classificar_tweet(row):
    if row['resposta'] == 'Considero misógina' and row['counts'] >= 3:
        return 'É Misógina'
    else:
        return 'Não misógina'

tweet_resposta_counts['classificação'] = tweet_resposta_counts.apply(classificar_tweet, axis=1)

tabela_final = tweet_resposta_counts[['Tweet', 'classificação']].drop_duplicates(subset=['Tweet'])

tabela_final.to_csv('tabela_final.csv', index=False)

tabela_final
```
